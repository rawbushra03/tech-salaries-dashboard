# Tech Salaries Exploratory Data Analysis (EDA)

This notebook explores the global tech salary dataset to understand trends, regional differences, and the impact of remote work.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv('../data/salaries_cleaned.csv')
df.head()
```

## 1. General Salary Distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['salary_in_usd'], bins=30, kde=True)
plt.title('Global Tech Salary Distribution')
plt.show()
```

## 2. Remote vs On-site

```python
df.groupby('remote_status')['salary_in_usd'].describe()
```

## 3. Top Roles by Salary

```python
top_roles = df.groupby('job_title')['salary_in_usd'].mean().sort_values(ascending=False).head(10)
top_roles.plot(kind='barh', title='Top 10 High Paying Roles')
plt.show()
```
